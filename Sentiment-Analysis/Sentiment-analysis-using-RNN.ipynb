```python
from keras import Sequential
from keras.layers import Dense, SimpleRNN, Embedding

# Define the vocabulary size (num_words used in imdb.load_data) and maxlen
vocab_size = 10000 # Default num_words for imdb.load_data
maxlen = 50        # Used for pad_sequences

# Redefine the model with an Embedding layer
model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=32, input_length=maxlen)) # Embedding layer
model.add(SimpleRNN(32, return_sequences=False)) # SimpleRNN layer, input_shape now inferred
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(loss = 'binary_crossentropy',optimizer = 'adam',metrics = ['accuracy'])
```

```python
history = model.fit(X_train,y_train,epochs = 5,validation_data = (X_test,y_test))
```

```python
from keras.datasets import imdb
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import pad_sequences
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data(num_words=10000)
```

```python
X_train = pad_sequences(X_train, padding='post', maxlen=50)
X_test = pad_sequences(X_test, padding='post', maxlen=50)
```

```python
X_train.shape
```

```python
model = Sequential()

model.add(Embedding(input_dim = 10000,output_dim = 2,input_length = 50))
model.add(SimpleRNN(32,return_sequences = False))
model.add(Dense(1,activation = 'sigmoid'))
model.summary()
```

```python
model.compile(loss = 'binary_crossentropy',optimizer = 'adam',metrics = ['accuracy'])
```

```python
history = model.fit(X_train,y_train,epochs = 5,validation_data = (X_test,y_test))
```
